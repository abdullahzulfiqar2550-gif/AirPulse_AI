```python
import pandas as pd
import matplotlib.pyplot as plt

# Data load karein
df = pd.read_csv(r"..\raw_data\pakistan_20261023_073030\pakistan_pm25_transformed.csv")
print("Total rows:", len(df))
```

```python
# Simple Histogram
plt.hist(df["pm25_value"].dropna(), bins=50, color="blue")
plt.title("PM2.5 Distribution")
plt.xlabel("PM2.5 Value")
plt.ylabel("Count")
plt.show()
```

```python
# Har mahine ka average PM2.5
monthly = df.groupby("month")["pm25_value"].mean()

plt.bar(monthly.index, monthly.values, color="orange")
plt.title("Monthly Average PM2.5")
plt.xlabel("Month")
plt.ylabel("Average PM2.5")
plt.show()
```

```python
# Har ghante (0 to 23) ka average
hourly = df.groupby("hour")["pm25_value"].mean()

plt.plot(hourly.index, hourly.values, marker="o", color="green")
plt.title("Hourly PM2.5 (0 to 23 Hours)")
plt.xlabel("Hour of Day")
plt.ylabel("Average PM2.5")
plt.show()
```

```python
# Har location ka average PM2.5
locations = df.groupby("location_name")["pm25_value"].mean().head(10)

plt.barh(locations.index, locations.values, color="red")
plt.title("Top Locations PM2.5")
plt.xlabel("Average PM2.5")
plt.show()
```

```python
# Simple Boxplot
plt.boxplot(df["pm25_value"].dropna())
plt.title("PM2.5 Box Plot")
plt.ylabel("PM2.5 Value")
plt.show()
```

```python
# Hafte ke 7 dino ka average (0=Mon, 6=Sun)
days = df.groupby("day_of_week")["pm25_value"].mean()

plt.bar(days.index, days.values, color="purple")
plt.title("PM2.5 by Day of Week")
plt.xlabel("Day (0=Monday, 6=Sunday)")
plt.ylabel("Average PM2.5")
plt.show()
```
